## Passo 1 — O que é um modelo conceitual?

Antes de criar qualquer tabela em um banco de dados, é preciso **planejar** o que será armazenado. Esse planejamento é chamado de **modelo conceitual** (ou **modelo Entidade-Relacionamento**, também chamado de **DER** — Diagrama Entidade-Relacionamento).

O modelo conceitual descreve, de forma visual e independente de tecnologia:

- **Entidades**: as "coisas" ou "objetos" do mundo real que queremos representar (ex.: `Estudante`, `Turma`, `Componente Curricular`).
- **Atributos**: as características de cada entidade (ex.: um `Estudante` tem `código do estudante`, `data de nascimento fictícia`).
- **Chave primária**: o atributo (ou conjunto de atributos) que identifica **de forma única** cada ocorrência de uma entidade (ex.: `codigo_estudante`).
- **Relacionamentos**: as conexões entre entidades (ex.: um `Estudante` **está matriculado em** uma `Turma`).
- **Cardinalidade**: quantas ocorrências de uma entidade se relacionam com quantas ocorrências de outra.
- **Participação**: se toda ocorrência de uma entidade **precisa** (obrigatória) ou **não precisa** (opcional) participar do relacionamento.

> 💬 **Analogia:** o modelo conceitual é como a **planta de uma casa** antes de construí-la. Antes de "erguer as paredes" (criar as tabelas no banco de dados), desenhamos como os cômodos (entidades) se conectam.

## Passo 2 — Entidades, atributos e chave primária

Vamos usar como exemplo um recorte da nossa base escolar fictícia. Observe a entidade abaixo:

### 🧩 Entidade: `Estudante`

| Atributo | Exemplo fictício | Observação |
|---|---|---|
| **codigo_estudante** (PK) | `EST001` | Chave primária — identifica cada estudante de forma única |
| nome_ficticio | `Estudante Alfa` | Nome genérico, sem dado real |
| turma_codigo | `T01` | Liga o estudante a uma turma (chave estrangeira) |

- A **chave primária** (PK, do inglês *primary key*) é marcada para deixar claro que **não pode se repetir** e **não pode ficar vazia** — é o "RG" daquela entidade dentro do banco de dados.
- Um atributo que aponta para a chave primária de **outra** entidade é chamado de **chave estrangeira** (FK, *foreign key*) — é o que permitirá, mais adiante, fazer os `JOIN`s entre tabelas.

### 🧩 Outras entidades do nosso cenário

### 📖 Glossário simples — Entidades e atributos

| Termo | Explicação simples |
|---|---|
| **Entidade** | Uma "coisa" do mundo real que queremos guardar informações (ex.: Estudante, Turma). |
| **Atributo** | Uma característica da entidade (ex.: nome, código). |
| **Chave primária (PK)** | O atributo que identifica cada registro de forma única, sem repetição. |
| **Chave estrangeira (FK)** | Um atributo que "aponta" para a chave primária de outra entidade, criando o relacionamento. |
| **Relacionamento** | A ligação/associação entre duas (ou mais) entidades. |

## Passo 3 — Cardinalidade dos relacionamentos

**Cardinalidade** indica **quantas** ocorrências de uma entidade podem se relacionar com **quantas** ocorrências de outra entidade. As três cardinalidades mais comuns são:

| Cardinalidade | Significado | Exemplo no nosso cenário escolar |
|---|---|---|
| **1:1** (um para um) | Uma ocorrência de A se relaciona com **no máximo uma** ocorrência de B, e vice-versa. | Um `Estudante` possui **um único** cartão de identificação escolar; cada cartão pertence a **um único** estudante. |
| **1:N** (um para muitos) | Uma ocorrência de A se relaciona com **várias** ocorrências de B, mas cada B se relaciona com **apenas uma** A. | Uma `Turma` possui **vários** `Estudantes`, mas cada `Estudante` pertence a **uma única** `Turma`. |
| **N:N** (muitos para muitos) | Várias ocorrências de A se relacionam com várias ocorrências de B. | Um `Estudante` pode cursar **vários** `Componentes`, e cada `Componente` é cursado por **vários** `Estudantes` (por isso existe a entidade associativa `Matrícula`, que "desmonta" o N:N em dois relacionamentos 1:N). |

> 💬 **Por que a Matrícula existe?** Bancos de dados relacionais não representam diretamente um relacionamento N:N — por isso criamos uma **entidade associativa** (`Matrícula`/`Resultado`) que guarda a combinação `codigo_estudante` + `codigo_componente`, transformando o N:N em dois relacionamentos 1:N.

## Passo 4 — Participação: obrigatória ou opcional?

Além da cardinalidade, cada entidade tem um tipo de **participação** no relacionamento:

- **Participação obrigatória (total)**: toda ocorrência da entidade **precisa** participar do relacionamento. Ex.: toda `Matrícula` **precisa** ter um `Estudante` associado — não existe matrícula "solta".
- **Participação opcional (parcial)**: pode existir uma ocorrência da entidade que **não participa** do relacionamento. Ex.: pode existir um `Estudante` que **ainda não** tem nenhuma `Matrícula` registrada em um componente (por exemplo, acabou de ingressar na escola).

> 🔑 **Conexão direta com o Módulo 2:** é exatamente essa **participação opcional** que vamos explorar com o `LEFT JOIN` — quando queremos listar **todos** os estudantes, mesmo aqueles que ainda **não têm** correspondência na tabela de matrículas/resultados.

### 📖 Glossário simples — Cardinalidade e participação

| Termo | Explicação simples |
|---|---|
| **Cardinalidade 1:1** | Uma ocorrência liga-se a, no máximo, uma outra ocorrência. |
| **Cardinalidade 1:N** | Uma ocorrência liga-se a várias outras, mas cada uma dessas se liga a apenas uma. |
| **Cardinalidade N:N** | Várias ocorrências de um lado se ligam a várias do outro lado. |
| **Entidade associativa** | Tabela criada para "resolver" um relacionamento N:N (ex.: Matrícula). |
| **Participação obrigatória** | Toda ocorrência da entidade precisa estar no relacionamento. |
| **Participação opcional** | A ocorrência pode existir sem participar do relacionamento. |

# 🟩 AULA 2 (EEBFM) / AULA 2 (EEJC) — Módulo 2: Junções Externas (LEFT JOIN)

**Defasagem trabalhada:** recuperar dados mesmo quando não há correspondência entre tabelas.

## Passo 1 — Relembrando o `INNER JOIN`

O `INNER JOIN` (junção interna) retorna **apenas as linhas que têm correspondência em ambas as tabelas**. Se um estudante não tem nenhuma matrícula, ele **não aparece** no resultado.

Vamos ver isso na prática: uma consulta juntando `estudantes` e `matriculas` usando `INNER JOIN`.

````python
# Consulta com INNER JOIN — só aparecem estudantes que têm ao menos uma matrícula
consulta_inner = '''
SELECT
    estudantes.codigo_estudante,
    estudantes.apelido_ficticio,
    matriculas.codigo_componente,
    matriculas.resultado
FROM estudantes
INNER JOIN matriculas ON estudantes.codigo_estudante = matriculas.codigo_estudante
ORDER BY estudantes.codigo_estudante
'''

resultado_inner = pd.read_sql_query(consulta_inner, conexao)
print("Quantidade de linhas no resultado do INNER JOIN:", len(resultado_inner))
resultado_inner
````

## Passo 2 — O que é uma junção externa (`LEFT JOIN`)?

Uma **junção externa** (*outer join*) é usada quando queremos manter **todas** as linhas de uma tabela, **mesmo que não haja correspondência** na outra tabela. No SQLite, a forma mais usada e mais compatível é o `LEFT JOIN` (junção externa à esquerda):

- `LEFT JOIN` mantém **todas as linhas da tabela à esquerda** (a primeira mencionada no `FROM`), preenchendo com `NULL` (nulo) as colunas da tabela à direita quando **não há correspondência**.

> ⚠️ **Nota técnica importante:** o padrão SQL também define `RIGHT JOIN` (junção externa à direita) e `FULL JOIN` (junção externa completa), mas **esses comandos têm diferenças de suporte entre os SGBDs**. O SQLite, por exemplo, historicamente **não oferecia suporte nativo a `RIGHT JOIN`/`FULL JOIN`** (isso vem mudando em versões recentes, mas o suporte não é universal). Por isso, **neste notebook focamos no `LEFT JOIN`**, que funciona de forma consistente em praticamente todos os SGBDs, incluindo o SQLite. Na prática, um `RIGHT JOIN` pode sempre ser reescrito como um `LEFT JOIN` apenas invertendo a ordem das tabelas.

## Passo 3 — Comparando `INNER JOIN` e `LEFT JOIN` na prática

Vamos repetir a mesma consulta do Passo 1, mas agora usando `LEFT JOIN`.

````python
# Consulta com LEFT JOIN — mantém TODOS os estudantes, mesmo sem matrícula correspondente
consulta_left = '''
SELECT
    estudantes.codigo_estudante,
    estudantes.apelido_ficticio,
    matriculas.codigo_componente,
    matriculas.resultado
FROM estudantes
LEFT JOIN matriculas ON estudantes.codigo_estudante = matriculas.codigo_estudante
ORDER BY estudantes.codigo_estudante
'''

resultado_left = pd.read_sql_query(consulta_left, conexao)
print("Quantidade de linhas no resultado do LEFT JOIN:", len(resultado_left))
resultado_left
````

## Passo 4 — Identificando os registros sem correspondência

Repare que o `EST010` aparece no resultado do `LEFT JOIN`, mas com `None`/`NaN` nas colunas de `matriculas` — isso mostra visualmente que ele **não tem nenhuma matrícula** cadastrada ainda.

Podemos filtrar exatamente essas linhas usando `WHERE ... IS NULL` no SQL, ou o método `.isna()` do pandas.

````python
# Filtrando, via SQL, os estudantes SEM nenhuma matrícula correspondente
consulta_sem_matricula = '''
SELECT
    estudantes.codigo_estudante,
    estudantes.apelido_ficticio
FROM estudantes
LEFT JOIN matriculas ON estudantes.codigo_estudante = matriculas.codigo_estudante
WHERE matriculas.codigo_matricula IS NULL
'''

resultado_sem_matricula = pd.read_sql_query(consulta_sem_matricula, conexao)
print("Estudantes sem nenhuma matrícula registrada:")
resultado_sem_matricula
````

## Passo 5 — Comparação visual dos resultados

| Aspecto | `INNER JOIN` | `LEFT JOIN` |
|---|---|---|
| **Linhas retornadas** | Só as que têm correspondência nas duas tabelas | Todas da tabela à esquerda, mesmo sem correspondência |
| **Valores sem par** | Não aparecem | Aparecem como `NULL`/`NaN` |
| **Uso típico** | "Quero só quem realmente tem X" | "Quero todos, e quero ENXERGAR quem não tem X" |
| **Neste exemplo** | 13 linhas (sem `EST010`) | 14 linhas (`EST010` aparece com valores nulos) |

> 💬 **Pergunta para reflexão:** em qual situação da escola faria mais sentido usar `LEFT JOIN` em vez de `INNER JOIN`? (ex.: relatório para identificar quem **ainda não** fez uma prova, quem **ainda não** entregou um trabalho...)

### 📖 Glossário simples — Módulo 2

| Termo | Explicação simples |
|---|---|
| **JOIN** | Comando SQL que combina linhas de duas (ou mais) tabelas com base em uma condição. |
| **INNER JOIN** | Só retorna linhas que existem em ambas as tabelas. |
| **LEFT JOIN (junção externa)** | Retorna todas as linhas da tabela à esquerda, mesmo sem correspondência na tabela à direita. |
| **NULL** | Valor "vazio"/desconhecido, usado quando não há correspondência. |
| **Tabela à esquerda** | A primeira tabela citada, logo após o `FROM`. |

# 🟨 AULA 3 (EEJC) — Módulo 3: Junções Múltiplas (três ou mais tabelas)

*(Na EEBFM, este módulo é trabalhado na mesma Aula 2, logo em seguida ao Módulo 2, de forma integrada.)*

**Defasagem trabalhada:** elaborar consultas combinando mais de duas tabelas.

## Passo 1 — Por que juntar mais de duas tabelas?

Em situações reais, a informação que queremos **não está toda em uma única tabela** — ela está espalhada em várias. Por exemplo, para responder "qual o **nome do componente** e o **resultado** de cada **estudante**, junto com o **nome da turma** dele?", precisamos combinar **quatro** tabelas: `estudantes`, `turmas`, `matriculas` e `componentes`.

## Passo 2 — Usando aliases (apelidos de tabela)

Quando o número de tabelas aumenta, os nomes completos (`estudantes.codigo_estudante`) deixam a consulta longa. Por isso, usamos **aliases** (apelidos curtos) para cada tabela, definidos logo após o nome dela no `FROM`/`JOIN`:

```sql
FROM estudantes AS e
JOIN turmas AS t ON e.codigo_turma = t.codigo_turma
```

A partir daí, usamos `e.` e `t.` no lugar dos nomes completos.

## Passo 3 — Encadeando múltiplos `JOIN`s

Uma junção múltipla nada mais é do que **vários `JOIN`s em sequência**, cada um ligando um novo par de tabelas.

````python
# Consulta com JUNÇÃO MÚLTIPLA — combina estudantes, turmas, matriculas e componentes (4 tabelas)
consulta_multipla = '''
SELECT
    e.codigo_estudante,
    e.apelido_ficticio,
    t.nome_turma,
    c.nome_componente,
    m.resultado
FROM estudantes AS e
JOIN turmas AS t ON e.codigo_turma = t.codigo_turma
JOIN matriculas AS m ON e.codigo_estudante = m.codigo_estudante
JOIN componentes AS c ON m.codigo_componente = c.codigo_componente
ORDER BY t.nome_turma, e.codigo_estudante, c.nome_componente
'''

resultado_multiplo = pd.read_sql_query(consulta_multipla, conexao)
print("Total de linhas no resultado:", len(resultado_multiplo))
resultado_multiplo
````

> 💬 Repare que usamos `JOIN` (equivalente a `INNER JOIN`) entre `estudantes`/`matriculas`/`componentes`, pois aqui queremos apenas quem **realmente tem** matrícula. Se quiséssemos manter também o `EST010` (sem matrícula) neste relatório combinado, trocaríamos o `JOIN` com `matriculas` por `LEFT JOIN` — os dois conceitos (Módulo 2 e Módulo 3) podem ser **combinados**!

## Passo 4 — Selecionando colunas específicas e ordenando o resultado

Em uma junção múltipla, é comum **selecionar apenas as colunas necessárias** (em vez de `SELECT *`) e **ordenar** o resultado com `ORDER BY` para facilitar a leitura — como já fizemos acima, ordenando por turma, depois por estudante, depois por componente.

````python
# Exemplo: junção múltipla combinando LEFT JOIN (Módulo 2) + múltiplas tabelas (Módulo 3)
# Mantém TODOS os estudantes (mesmo sem matrícula), combinando com turma e, quando existir, componente/resultado
consulta_multipla_com_left = '''
SELECT
    e.codigo_estudante,
    e.apelido_ficticio,
    t.nome_turma,
    c.nome_componente,
    m.resultado
FROM estudantes AS e
JOIN turmas AS t ON e.codigo_turma = t.codigo_turma
LEFT JOIN matriculas AS m ON e.codigo_estudante = m.codigo_estudante
LEFT JOIN componentes AS c ON m.codigo_componente = c.codigo_componente
ORDER BY t.nome_turma, e.codigo_estudante
'''

resultado_multiplo_left = pd.read_sql_query(consulta_multipla_com_left, conexao)
print("Total de linhas (deve incluir EST010 com valores nulos):", len(resultado_multiplo_left))
resultado_multiplo_left[resultado_multiplo_left["codigo_estudante"] == "EST010"]
````

### 📖 Glossário simples — Módulo 3

| Termo | Explicação simples |
|---|---|
| **Junção múltipla** | Consulta que combina três ou mais tabelas usando vários `JOIN`s em sequência. |
| **Alias (apelido de tabela)** | Nome curto dado a uma tabela (ex.: `AS e`) para simplificar a escrita da consulta. |
| **ORDER BY** | Cláusula que define a ordem de exibição das linhas do resultado. |
| **SELECT (colunas específicas)** | Escolher apenas as colunas necessárias, em vez de trazer todas com `*`. |

consulta_aprofundamento = '''
SELECT
    o.nome_oficina,
    AVG(i.frequencia_percentual) AS frequencia_media
FROM oficinas AS o
JOIN inscricoes AS i ON o.codigo_oficina = i.codigo_oficina
GROUP BY o.nome_oficina
ORDER BY frequencia_media DESC
'''
